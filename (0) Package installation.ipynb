#### Please copy and paste the following code in R or RStudio console for installation. 

```r
# -------------------------------
# Install required package managers
# -------------------------------
if (!requireNamespace("BiocManager", quietly = TRUE))
  install.packages("BiocManager")

# -------------------------------
# Define package lists
# -------------------------------

# CRAN packages
cran_packages <- c(
  "dplyr", "Seurat", "patchwork", "stringr", "ggplot2",
  "hdf5r", "repr", "RColorBrewer", "tidyverse",
  "DoubletFinder", "RCurl", "cowplot", "clustree",
  "gridExtra", "HGNChelper", "openxlsx", "igraph",
  "data.tree", "ggraph", "viridis", "fedmatch",
  "EnhancedVolcano", "msigdbr", "fgsea", "readxl",
  "plotly", "shiny", "tibble", "pheatmap", "tidyr"
)

# Bioconductor packages
bioc_packages <- c(
  "scDblFinder",
  "BiocParallel"
)

# -------------------------------
# Install missing CRAN packages
# -------------------------------
installed_cran <- rownames(installed.packages())
for (pkg in cran_packages) {
  if (!pkg %in% installed_cran) {
    install.packages(pkg, dependencies = TRUE)
  }
}

# -------------------------------
# Install missing Bioconductor packages
# -------------------------------
installed_bioc <- rownames(installed.packages())
for (pkg in bioc_packages) {
  if (!pkg %in% installed_bioc) {
    BiocManager::install(pkg, ask = FALSE)
  }
}
```
